# microWakeWord — Custom wake word training

Full pipeline: test pronunciations, generate samples, train, export `.tflite`.

## Workarounds applied
- `datasets==2.21.0` (newer versions require torchcodec which breaks audio decoding)
- `piper-tts` needed for piper-sample-generator
- Augmentation data downloaded via soundfile/pydub (bypasses datasets Audio decoder)
- microWakeWord `utils.py` patched: stride assertion replaced with truncation
- `clip_duration_ms` must produce frames divisible by stride (3). Use 1500 or 1800, NOT 2000
- `dinner_party_eval` dataset removed from training config (causes OOM on Colab free tier)

```python
# Cell 5: Test pronunciation variants (listen to all candidates)
# 1. Run this cell to hear each variant
# 2. Listen to the audio samples
# 3. In the next cell (5b), type the numbers of the ones you want to keep and press Enter

import os, json, sys, shutil, urllib.request
from pathlib import Path

local_g2pw = Path("/content/piper-sample-generator/g2pW")
os.makedirs(local_g2pw, exist_ok=True)

# Direct raw download links bypassing Hugging Face Git LFS redirects
resources = {
    'g2pw.onnx': 'https://huggingface.co/datasets/rhasspy/piper-checkpoints/resolve/main/zh/zh_CN/_resources/g2pw.onnx?download=true',
    'binyin_g2pw.npz': 'https://huggingface.co/datasets/rhasspy/piper-checkpoints/resolve/main/zh/zh_CN/_resources/binyin_g2pw.npz?download=true',
    'char_to_pinyin.json': 'https://huggingface.co/datasets/rhasspy/piper-checkpoints/resolve/main/zh/zh_CN/_resources/char_to_pinyin.json?download=true',
    'pinyin_to_char.json': 'https://huggingface.co/datasets/rhasspy/piper-checkpoints/resolve/main/zh/zh_CN/_resources/pinyin_to_char.json?download=true',
    'bopomofo_to_pinyin_wo_tune_dict.json': 'https://huggingface.co/datasets/rhasspy/piper-checkpoints/resolve/main/zh/zh_CN/_resources/bopomofo_to_pinyin_wo_tune_dict.json?download=true'
}

# Rigorously validate and download clean assets
for fname, url in resources.items():
    target_file = local_g2pw / fname

    # Always force validate existing JSON files to prevent hidden HTML/LFS corrupted files
    if target_file.exists():
        is_corrupt = False
        if fname.endswith('.json'):
            try:
                with open(target_file, 'r') as f:
                    json.load(f)
            except Exception:
                print(f"Detected corrupted JSON file for {fname}. Purging...")
                is_corrupt = True
        elif target_file.stat().st_size < 1000:  # Too small for model files
            print(f"Detected suspicious undersized model file for {fname}. Purging...")
            is_corrupt = True

        if is_corrupt:
            target_file.unlink(missing_ok=True)

    # Download if not present or was purged
    if not target_file.exists():
        print(f"Downloading clean resource: {fname}...")
        try:
            # Use curl -L to follow redirects reliably
            !curl -L -s -o "{target_file}" "{url}"
        except Exception as e:
            print(f"Failed to download {fname}: {e}")

try:
    import piper
    piper_dir = Path(piper.__file__).parent
    target_g2pw = piper_dir / "g2pW"
    os.makedirs(target_g2pw, exist_ok=True)

    # Copy verified files to all potential search locations
    for fname in resources.keys():
        src_file = local_g2pw / fname
        if src_file.exists():
            shutil.copy2(src_file, target_g2pw / fname)
            shutil.copy2(src_file, Path("/content/piper-sample-generator/g2pW") / fname)
            shutil.copy2(src_file, piper_dir / fname)
except Exception as e:
    print(f"Warning during environment copying: {e}")

# Restore original __main__.py file from git to ensure a clean start without corrupted indentation
generator_main = Path("/content/piper-sample-generator/piper_sample_generator/__main__.py")
if generator_main.exists():
    !git -C /content/piper-sample-generator checkout -- piper_sample_generator/__main__.py

    with open(generator_main, "r") as f:
        code_lines = f.readlines()

    # Correctly aligned and unindented patch codes
    patch_code = (
        "\n"
        "import pathlib\n"
        "import piper.g2pw_onnx\n"
        "piper.g2pw_onnx._find_data_dir = lambda *args: pathlib.Path('/content/piper-sample-generator/g2pW')\n"
    )

    wave_patch_code = (
        "\n"
        "import wave\n"
        "original_wave_open = wave.open\n"
        "def patched_wave_open(*args, **kwargs):\n"
        "    obj = original_wave_open(*args, **kwargs)\n"
        "    if hasattr(obj, '_ensure_header_written'):\n"
        "        original_close = obj.close\n"
        "        def patched_close():\n"
        "            try:\n"
        "                if obj._nchannels == 0: obj.setnchannels(1)\n"
        "                if obj._sampwidth == 0: obj.setsampwidth(2)\n"
        "                if obj._framerate == 0: obj.setframerate(16000)\n"
        "            except:\n"
        "                pass\n"
        "            original_close()\n"
        "        obj.close = patched_close\n"
        "    return obj\n"
        "wave.open = patched_wave_open\n"
    )

    # Inject patches right after the first line (shebang)
    code_lines.insert(1, patch_code)
    code_lines.insert(2, wave_patch_code)

    with open(generator_main, "w") as f:
        f.writelines(code_lines)
    print("Successfully patched piper_sample_generator with clean syntax indentation.")

from IPython.display import Audio, display, HTML

with open('/content/config.json') as f:
    CONFIG = json.load(f)

VOICES = CONFIG['VOICES']
MODEL = f'/content/{list(VOICES.keys())[0]}.onnx'

for i, word in enumerate(CONFIG['TEST_VARIANTS']):
    slug = word.replace(' ', '_')
    out_dir = f'/content/test_{slug}'
    os.makedirs(out_dir, exist_ok=True)

    !cd /content/piper-sample-generator && python -m piper_sample_generator \
        "{word}" \
        --model {MODEL} \
        --max-samples 1 \
        --output-dir {out_dir}

    display(HTML(f'<h3>#{i}: {word}</h3>'))
    wav_file = f'{out_dir}/0.wav'
    if os.path.exists(wav_file):
        display(Audio(wav_file))
    else:
        print(f'Error: Failed to generate audio for "{word}". Please check traceback above.')
```

```python
# Cell 7: Download augmentation data (MIT RIRs, AudioSet, FMA)

import soundfile as sf
import scipy
import os
import numpy as np
from pathlib import Path
from tqdm import tqdm
from datasets import load_dataset, Audio
import io

# MIT Room Impulse Responses
output_dir = './mit_rirs'
if not os.path.exists(output_dir) or len(os.listdir(output_dir)) == 0:
    os.makedirs(output_dir, exist_ok=True)
    rir_dataset = load_dataset('davidscripka/MIT_environmental_impulse_responses', split='train')
    rir_dataset = rir_dataset.cast_column('audio', Audio(decode=False))
    for i, row in enumerate(tqdm(rir_dataset)):
        audio_path = row['audio']['path']
        if audio_path and os.path.exists(audio_path):
            data, sr = sf.read(audio_path)
            if len(data.shape) > 1:
                data = data[:, 0]
            if sr != 16000:
                import librosa
                data = librosa.resample(data, orig_sr=sr, target_sr=16000)
            scipy.io.wavfile.write(os.path.join(output_dir, f'rir_{i}.wav'), 16000, (data*32767).astype(np.int16))

# AudioSet
if not os.path.exists('audioset'):
    os.mkdir('audioset')
    fname = 'bal_train09.tar'
    link = 'https://huggingface.co/datasets/agkphysics/AudioSet/resolve/main/data/' + fname
    !wget -O audioset/{fname} {link}
    !cd audioset && tar -xf bal_train09.tar

output_dir = './audioset_16k'
if not os.path.exists(output_dir) or len(os.listdir(output_dir)) == 0:
    os.makedirs(output_dir, exist_ok=True)
    for flac in tqdm(list(Path('audioset/audio').glob('**/*.flac'))):
        try:
            data, sr = sf.read(str(flac))
            if len(data.shape) > 1:
                data = data[:, 0]
            if sr != 16000:
                import librosa
                data = librosa.resample(data, orig_sr=sr, target_sr=16000)
            scipy.io.wavfile.write(os.path.join(output_dir, flac.stem + '.wav'), 16000, (data*32767).astype(np.int16))
        except Exception:
            pass

# FMA
if not os.path.exists('fma') or not os.path.exists('fma/fma_small'):
    os.makedirs('fma', exist_ok=True)
    fname = 'fma_xs.zip'
    link = 'https://huggingface.co/datasets/mchl914/fma_xsmall/resolve/main/' + fname
    !wget -O fma/{fname} {link}
    !cd fma && unzip -q {fname}

output_dir = './fma_16k'
if not os.path.exists(output_dir) or len(os.listdir(output_dir)) == 0:
    os.makedirs(output_dir, exist_ok=True)
    from pydub import AudioSegment
    for mp3 in tqdm(list(Path('fma/fma_small').glob('**/*.mp3'))):
        try:
            audio = AudioSegment.from_mp3(str(mp3)).set_frame_rate(16000).set_channels(1)
            samples = np.array(audio.get_array_of_samples(), dtype=np.int16)
            scipy.io.wavfile.write(os.path.join(output_dir, mp3.stem + '.wav'), 16000, samples)
        except Exception:
            pass

print('Augmentation data ready')
```

```python
# Cell 8: Generate spectrogram features

try:
    from microwakeword.audio.augmentation import Augmentation
    from microwakeword.audio.clips import Clips
    from microwakeword.audio.spectrograms import SpectrogramGeneration
except ModuleNotFoundError:
    raise RuntimeError('microWakeWord not installed. Re-run from Cell 2 (VM was reset).')

from mmap_ninja.ragged import RaggedMmap
import os

clips = Clips(
    input_directory='generated_samples',
    file_pattern='*.wav',
    max_clip_duration_s=None,
    remove_silence=False,
    random_split_seed=10,
    split_count=0.1,
)

augmenter = Augmentation(
    augmentation_duration_s=3.2,
    augmentation_probabilities={
        'SevenBandParametricEQ': 0.1,
        'TanhDistortion': 0.1,
        'PitchShift': 0.1,
        'BandStopFilter': 0.1,
        'AddColorNoise': 0.1,
        'AddBackgroundNoise': 0.75,
        'Gain': 1.0,
        'RIR': 0.5,
    },
    impulse_paths=['mit_rirs'],
    background_paths=['fma_16k', 'audioset_16k'],
    background_min_snr_db=-5,
    background_max_snr_db=10,
    min_jitter_s=0.195,
    max_jitter_s=0.205,
)

output_dir = 'generated_augmented_features'
os.makedirs(output_dir, exist_ok=True)

for split in ['training', 'validation', 'testing']:
    out_dir = os.path.join(output_dir, split)
    os.makedirs(out_dir, exist_ok=True)

    split_name = {'training': 'train', 'validation': 'validation', 'testing': 'test'}[split]
    repetition = 2 if split == 'training' else 1
    slide = 1 if split == 'testing' else 10

    spectrograms = SpectrogramGeneration(
        clips=clips,
        augmenter=augmenter,
        slide_frames=slide,
        step_ms=10,
    )

    RaggedMmap.from_generator(
        out_dir=os.path.join(out_dir, 'wakeword_mmap'),
        sample_generator=spectrograms.spectrogram_generator(split=split_name, repeat=repetition),
        batch_size=100,
        verbose=True,
    )

print('Spectrogram features ready')
```
